#1st Step Data Cleaning

```python
#Checking Missing Values

#dataset.isnull().sum()
#dataset.isnull().any(axis=1) #(Thats mean findout any column have null values)
dataset[dataset.isnull().any(axis=1)] #(Thats mean findout directly those column have null values)

```

The dataset is converted into two sets based on Region from 122th index, we can make a new column based on the Region

1 : "Bejaia Region Dataset"

2 : "Sidi-Bel Abbes Region Dataset"

Add new column with region

```python
dataset.loc[:122, "Region"]=0
dataset.loc[122:, "Region"]=1
df=dataset
df.info()
```

```python
#converting Region into INT
df[['Region']]=df[['Region']].astype(int)
df.head()
```

```python
#Cheking Null Values
dataset.isnull().sum()
```

```python
#Dropping/Removing Null values Row
df=df.dropna().reset_index(drop=True)
```

```python
df.head()
```

```python
dataset.isnull().sum()
```

```python
df.iloc[['122']]
```

```python
df=df.drop(122).reset_index(drop=True)
```

```python
df.iloc[['122']]
```

```python
#fixing a spaces in columns names
df.columns
```

```python
df.columns=df.columns.str.strip()
df.columns
```

```python
df.info()
```

```python
#changes the required columns as Integer Datatype
df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']]=df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']].astype(int)
```

```python
df.info()
```

```python
#Changing the other Columns to float dataType (2ndway)
# Objects=[features for features in df.columns if df[features].dtypes=='O'] #For Older Version

# print(df.dtypes)
Object=[df.select_dtypes(include=['object', 'str']).columns.tolist()]

Objects=Object[0]
print(Objects)
```

```python
for i in Objects:
    if i != 'Classes':
        df[i] = df[i].astype(float)
```

```python
#Lets save the cleaned dataset
df.to_csv('../dataset/Algerian_forest_fires_cleaned_dataset.csv', index=False)
```
